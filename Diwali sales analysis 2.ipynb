## 💼 07 | OCCUPATION ANALYSIS

**Analyzing customer orders across different occupations.**


```python
sales_occupation = df.groupby('Occupation', as_index=False)['Orders'].sum().sort_values(
    by='Orders', ascending=False
)

plt.figure(figsize=(12,6))

sns.barplot(
    x='Occupation',
    y='Orders',
    data=sales_occupation,
    hue='Occupation',
    palette='husl',
    legend=False
)

plt.title('Occupation-wise Total Orders', fontsize=15, fontweight='bold')
plt.xlabel('Occupation')
plt.ylabel('Total Orders')
plt.xticks(rotation=45)
plt.grid(False)

plt.show()
```

<b>From the above graph we can say that the most of the buyers are working in IT, Aviation and Healthcare Center</b>

## 🛍️ 08 | PRODUCT CATEGORY ANALYSIS

**Identifying the most popular product categories based on customer demand.**


```python
sales_product = df.groupby(
    'Product_Category', as_index=False
)['Orders'].sum().sort_values(
    by='Orders', ascending=False
).head(10)

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    x='Orders',
    y='Product_Category',
    data=sales_product,
    hue='Product_Category',
    palette='husl',
    legend=False
)

# Values
for container in ax.containers:
    ax.bar_label(
        container,
        fontsize=10,
        fontweight='bold',
        padding=3
    )

plt.title(
    'Top 10 Product Categories by Orders',
    fontsize=15,
    fontweight='bold',
    pad=15
)

plt.xlabel('Total Orders', fontsize=11)
plt.ylabel('Product Category', fontsize=11)

plt.grid(False)
plt.tight_layout()
plt.show()
```
